```python
# Task 2: Exploratory Data Analysis

# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Load dataset
df = pd.read_csv('/content/Student_performance.csv')

# 3. Explore the dataset
print("First 5 Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

# 4. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 5. Check duplicate values
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# 6. Statistical summary
print("\nStatistical Summary:")
print(df.describe())

# 7. Average GPA
print("\nAverage GPA:")
print(df['GPA'].mean())

# 8. GPA Distribution
plt.figure(figsize=(6,4))
sns.histplot(df['GPA'], bins=20, kde=True)
plt.title('GPA Distribution')
plt.xlabel('GPA')
plt.ylabel('Number of Students')
plt.show()

# 9. Study Time vs GPA
plt.figure(figsize=(6,4))
sns.scatterplot(x='StudyTimeWeekly', y='GPA', data=df)
plt.title('Study Time vs GPA')
plt.xlabel('Study Time Weekly')
plt.ylabel('GPA')
plt.show()

# 10. Students by Age
plt.figure(figsize=(6,4))
sns.countplot(x='Age', data=df)
plt.title('Students by Age')
plt.xlabel('Age')
plt.ylabel('Number of Students')
plt.show()

# 11. GPA by Gender
plt.figure(figsize=(6,4))
sns.boxplot(x='Gender', y='GPA', data=df)
plt.title('GPA by Gender')
plt.xlabel('Gender')
plt.ylabel('GPA')
plt.show()

# 12. Correlation Heatmap
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()

# 13. Final Findings
print("\nEDA Findings:")
print("1. The dataset contains", df.shape[0], "students.")
print("2. The dataset contains", df.shape[1], "columns.")
print("3. Missing values:", df.isnull().sum().sum())
print("4. Duplicate rows:", df.duplicated().sum())
print("5. Average GPA:", df['GPA'].mean())
print("6. Maximum GPA:", df['GPA'].max())
print("7. Minimum GPA:", df['GPA'].min())
print("8. Average Study Time:", df['StudyTimeWeekly'].mean())
```
